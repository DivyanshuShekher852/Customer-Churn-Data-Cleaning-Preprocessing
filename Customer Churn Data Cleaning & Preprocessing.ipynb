## 2. Load the Dataset

The customer churn CSV file is loaded using Pandas.

The `read_csv()` function is used to read the raw dataset into a Pandas DataFrame.

```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\customer_churn_sample.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 4. Check Missing Values

Missing values can cause problems during analysis and machine learning.

We calculate missing values for every column and then calculate the total number of missing values in the dataset.

```python
missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

### Missing Value Treatment

If missing values are found, they should be handled according to the characteristics of the column.

Possible approaches include:

- Mean/median imputation for numerical columns
- Mode imputation for categorical columns
- Removing rows when appropriate
- Removing columns when a column contains excessive missing data

For this dataset, no missing values are present. Therefore, no imputation or row deletion is required.

## 5. Check Duplicate Records

Duplicate records can lead to incorrect analysis and misleading results.

The `duplicated()` function is used to identify duplicate rows.

```python
duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)
```

```python
if duplicate_count == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records found:")
    display(df[df.duplicated(keep=False)])
```

## 6. Standardize Column Headers

Column names should follow a consistent naming convention.

The following transformations are applied:

- Remove leading and trailing spaces
- Convert CamelCase into snake_case
- Convert all characters to lowercase
- Replace special characters with underscores
- Remove unnecessary underscores

This makes the dataset easier to work with in Python and SQL.

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(
        r"([a-z0-9])([A-Z])",
        r"\1_\2",
        regex=True
    )
    .str.lower()
    .str.replace(
        r"[^a-z0-9]+",
        "_",
        regex=True
    )
    .str.strip("_")
)

print("Standardized column names:")
print(df.columns.tolist())
```

## 7. Clean Categorical Columns

Categorical values may contain unwanted leading or trailing spaces.

For example:

`Male `

and

`Male`

could be treated as different values.

To prevent this problem, whitespace is removed from categorical columns.

```python
categorical_columns = df.select_dtypes(include=["object", "string"]).columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
# Cell 13 in current notebook:
categorical_columns = df.select_dtypes(include=["str", "object"]).columns
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include=["str", "object"]).columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
for column in categorical_columns:
    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
    )

print("Categorical values standardized successfully!")
```

## 8. Check and Correct Data Types

Correct data types are important for accurate analysis.

The following columns are expected to contain integer values:

- Age
- Tenure Months
- Support Tickets

The following columns are expected to contain decimal/numeric values:

- Monthly Charges
- Total Charges

```python
integer_columns = [
    "age",
    "tenure_months",
    "support_tickets"
]

numeric_columns = [
    "monthly_charges",
    "total_charges"
]

for column in integer_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="raise"
    ).astype("int64")

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="raise"
    ).astype("float64")

print("Data types corrected successfully!")
```

```python
print(df.dtypes)
```

## 9. Check Date Columns

Date columns should use a consistent date format.

We first identify columns containing words such as:

- `date`
- `time`

If date columns exist, they can then be converted using `pd.to_datetime()`.

The supplied customer churn dataset does not contain date/time columns, so date conversion is not required.

```python
date_columns = [
    column
    for column in df.columns
    if "date" in column.lower()
    or "time" in column.lower()
]

print("Date/Time columns:")

if date_columns:
    print(date_columns)
else:
    print("No date/time columns found.")
```

## 10. Validate Customer IDs

Customer IDs should be unique and should not contain missing values.

This check ensures that each customer can be uniquely identified.

```python
print(
    "Customer IDs are unique:",
    df["customer_id"].is_unique
)

print(
    "Missing Customer IDs:",
    df["customer_id"].isnull().sum()
)
```

## 11. Validate Numerical Data

Business rules are used to identify invalid numerical values.

The following conditions are checked:

- Age should be between 18 and 100.
- Tenure should not be negative.
- Monthly charges should not be negative.
- Total charges should not be negative.
- Support ticket count should not be negative.

```python
print(
    "Age valid:",
    df["age"].between(18, 100).all()
)

print(
    "Tenure valid:",
    df["tenure_months"].ge(0).all()
)

print(
    "Monthly charges valid:",
    df["monthly_charges"].ge(0).all()
)

print(
    "Total charges valid:",
    df["total_charges"].ge(0).all()
)

print(
    "Support tickets valid:",
    df["support_tickets"].ge(0).all()
)
```

## 12. Validate Total Charges

The dataset contains both monthly charges and tenure information.

As an additional quality check, we verify whether:

**Total Charges = Monthly Charges × Tenure Months**

A tolerance of `0.01` is used to account for decimal rounding.

```python
expected_total = (
    df["monthly_charges"]
    * df["tenure_months"]
).round(2)

total_charges_check = np.isclose(
    df["total_charges"],
    expected_total,
    atol=0.01
)

print(
    "Total charges consistent:",
    total_charges_check.all()
)
```

## 13. Final Data Quality Check

After completing the cleaning process, the dataset is checked again.

The final validation confirms:

- Number of rows
- Number of columns
- Missing values
- Duplicate records
- Unique customer IDs
- Valid numerical values
- Total charge consistency

```python
print("=" * 50)
print("FINAL DATA QUALITY REPORT")
print("=" * 50)

print("Final Rows:", df.shape[0])
print("Final Columns:", df.shape[1])

print(
    "Missing Values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate Records:",
    df.duplicated().sum()
)

print(
    "Customer IDs Unique:",
    df["customer_id"].is_unique
)

print(
    "Total Charges Consistent:",
    total_charges_check.all()
)
```

## 14. Export Clean Dataset

After completing all cleaning and validation steps, the cleaned DataFrame is exported as a CSV file.

The `index=False` parameter prevents Pandas from adding an unnecessary index column to the output file.

```python
df.to_csv(
    "customer_churn_clean.csv",
    index=False
)

print(
    "customer_churn_clean.csv "
    "exported successfully!"
)
```
